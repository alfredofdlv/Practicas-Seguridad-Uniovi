```python
import random

def alterar_mensaje_cifrado(mensaje_cifrado):
    """
    Toma un mensaje en bytes, cambia un byte aleatorio por un valor diferente,
    y devuelve el nuevo mensaje alterado.
    """
    # 1. Convertimos a bytearray porque los 'bytes' normales no se pueden modificar
    mensaje_mutable = bytearray(mensaje_cifrado)
    
    # 2. Elegimos una posición al azar dentro del mensaje
    # random.randint incluye ambos extremos, por eso restamos 1 a la longitud
    posicion_aleatoria = random.randint(0, len(mensaje_mutable) - 1)
    
    # 3. Guardamos el byte original para asegurarnos de que realmente lo cambiamos
    byte_original = mensaje_mutable[posicion_aleatoria]
    
    # 4. Generamos un nuevo byte (un número entre 0 y 255)
    nuevo_byte = random.randint(0, 255)
    
    # Nos aseguramos de que el nuevo byte sea diferente al original
    while nuevo_byte == byte_original:
        nuevo_byte = random.randint(0, 255)
        
    # 5. Aplicamos el cambio
    mensaje_mutable[posicion_aleatoria] = nuevo_byte
    
    print(f"[*] Ataque simulado: Byte en posición {posicion_aleatoria} cambiado de {byte_original} a {nuevo_byte}")
    
    # 6. Volvemos a convertir a bytes inmutables y lo devolvemos
    return bytes(mensaje_mutable)
```

# 4 Funcion `new()`
La  función new() devuelve  un  nuevo  objeto  de  la  clase  hash que  implementa  la  función hash que se le especifique. En este caso, el primer parámetro deberá ser una cadena con el nombre de la función hash que se quiera utilizar (“sha1”, “md5”, “sha256”, etc.) y el segundo parámetro cualquier tipo de cadena que queramos cifrar. 

Añade el uso de esta función a las pruebas realizadas anteriormente. Para ello, crea una variable a la que asignes una llamada a  la  función  new  pasándole  el  algoritmo “sha256”y  el  mensaje b”texto”.  

Imprime  esta variable utilizando el método digets() y hexdigets(). 

Cambia el algoritmo a otro de los conocidos y comprueba de nuevo la salida


```python
new_var=hashlib.new("sha256",b"texto")
display('Hash: ', new_var.digest())
display('Hash en hexadecimal: ', new_var.hexdigest())
```

# 5 Funcion `update()`

La   función `update()` actualizará   el   objeto   hash   añadiendo   nueva   información (normalmente otra  cadena).  Pese  a  que  se  realice  más  de  una  llamada  al  método, será  lo equivalente  a  realizar  una  única  llamada.  Utiliza  este  método  para  realizar  el  hashde  un mensaje  que  actualizarás  posteriormente.  Para  ello,  crea  una  variable  (mensaje)  y  asigna directamente  la  llamada  al  objeto hashlibusando  el  método `sha256()`.  Posteriormente utiliza la función `update()` 3 veces(ya no hace falta usar el objeto hashlib) para actualizar el valor de la variable mensaje creada anteriormente. Puedes introducir el texto que quieras, por ejemplo, “mensaje” en la primera actualización, “super” en la segunda y “secreto” en la tercera. Posteriormente, muestra por pantalla el mensaje

```python
mensaje=hashlib.sha256(b"mensaje")
print(mensaje.digest())
print(f'Len Hash: {mensaje.digest_size}', )
print(f'Len Hash en hexadecimal: {len(mensaje.hexdigest())}')


mensaje.update(b"super")
print(mensaje.digest())
print(f'Len Hash: {mensaje.digest_size}', )
print(f'Len Hash en hexadecimal: {len(mensaje.hexdigest())}')


mensaje.update(b"secreto")
print(mensaje.digest())
print(f'Len Hash: {len(mensaje.digest())}', )
print(f'Len Hash en hexadecimal: {len(mensaje.hexdigest())}')

```

```python
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import hashes
```

```python
digest=hashes.Hash(hashes.SHA256(),default_backend)

digest.update(b"Mensaje")
print('Mensaje',digest)
digest.update(b"Secreto")
print('Con secreto',digest)

print(digest.algorithm)

digest_final=digest.finalize().hex()
print(digest_final)
```
